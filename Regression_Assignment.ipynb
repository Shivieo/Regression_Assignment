```python
#5 How do we calculate the slope m in Simple Linear Regression?
# m = Σ[(xi - x̄)(yi - ȳ)] / Σ[(xi - x̄)²]
# where xi -> independent variable x
# x̄ -> average of independent variables
# yi -> dependent variable y
# ȳ -> average of dependent variable
# To find m, the steps are:
# first calculate mean of independent and dependent variable
# second, subtract the mean from their every independent and dependent variable
# third, multiply the differences of both to each other
# fourth, sum all the products values collectively
# fifth, divide it by the summation of the differences of the squares of independent variables.
```

```python
#18 What are the limitations of using R² as a sole measure of model performance?
# While R² is a commonly used metric to evaluate the performance of a regression model, it has several limitations when used as the sole measure:

# Assumes a linear relationship: R² measures the proportion of variance in the dependent variable that is explained by the independent variables, assuming a linear relationship. If the true relationship is non-linear, R² may not accurately reflect the goodness of fit.

# Sensitive to outliers: R² can be heavily influenced by outliers in the data. A few extreme observations can disproportionately inflate or deflate the R² value, leading to misleading conclusions about the model's performance.

# Cannot establish causality: A high R² value indicates a strong correlation between the variables but does not imply causation. There might be other underlying factors influencing the relationship, which R² does not capture.

# May not detect a biased model: A model can have a high R² value even if it is biased, meaning it consistently over- or under-predicts the dependent variable. R² does not provide information about the accuracy of predictions or potential biases in the model.

# Increases with the number of variables: Adding more independent variables to the model will always increase R², even if those variables are not truly related to the dependent variable. This makes it difficult to compare models with different numbers of predictors using R² alone.

# Limited applicability for non-linear models: R² is primarily designed for linear regression models. Applying it to non-linear models can be misleading as it does not accurately capture the goodness of fit in such cases.

# Doesn't measure predictive accuracy: While a high R² suggests a good fit to the data, it doesn't directly measure how well the model predicts new, unseen data. Other metrics like mean squared error or prediction intervals might be more informative in assessing predictive accuracy.
```

```python
#19  How would you interpret a large standard error for a regression coefficient?
# A large standard error for a regression coefficient indicates that the estimated coefficient is imprecise and unreliable.

# Implications of a Large Standard Error:
# Uncertainty about the true effect: A large standard error means that the true effect of the predictor variable on the response variable is estimated with
# considerable uncertainty. The coefficient estimate could be quite different from the true population value.
# Difficulty in establishing statistical significance: In hypothesis testing, a large standard error makes it harder to reject the null hypothesis (that the
# coefficient is zero). This is because the t-statistic (coefficient divided by standard error) will be smaller, leading to a higher p-value. If the p-value
# is above the chosen significance level (e.g., 0.05), you fail to reject the null hypothesis, meaning you don't have enough evidence to conclude that the
# predictor has a statistically significant effect.
# Wider confidence intervals: The confidence interval for the coefficient, which provides a range of plausible values for the true effect, will be wider with a
# larger standard error. This reflects the greater uncertainty in the estimate.
```

```python
#20 How can heteroscedasticity be identified in residual plots, and why is it important to address it?
# Identifying Heteroscedasticity in Residual Plots
# The most common way to visually detect heteroscedasticity is by examining a scatterplot of residuals against predicted (fitted) values. Here's what to look
# for:
# Funnel Shape: The most classic sign of heteroscedasticity is a funnel or cone shape in the residual plot. This means that the spread of residuals increases or
# decreases as the predicted values increase.
# Non-Constant Spread: Even if it's not a perfect funnel, look for any pattern where the vertical spread of the residuals changes noticeably across the range of
# predicted values. If the residuals seem to be more scattered on one side of the plot than the other, it suggests heteroscedasticity.
# No Obvious Pattern (Homoscedasticity): In contrast, if the residuals are randomly scattered around zero with roughly constant variance across all predicted
# values, it indicates homoscedasticity (constant variance), which is what we ideally want.
```

```python
#26 What is the general equation for polynomial regression?
# The general equation for polynomial regression is:

# y = β₀ + β₁x + β₂x² + β₃x³ + ... + βₙxⁿ + ε

# Where:

# y: The dependent variable (the variable you're trying to predict).
# x: The independent variable (the variable you're using to make predictions).
# β₀, β₁, β₂, β₃, ..., βₙ: The coefficients of the polynomial. These are the values the model learns during the regression process.
# β₀ is the y-intercept (the value of y when x = 0).
# β₁ is the coefficient for the linear term (x).
# β₂ is the coefficient for the quadratic term (x²).
# β₃ is the coefficient for the cubic term (x³).
# And so on...
#
# n: The degree of the polynomial. This is the highest power of x in the equation. It determines the complexity of the curve.
# ε: The error term (also called the residual). This represents the difference between the observed values of y and the values predicted by the model. It
#    accounts for the variability in y that is not explained by the polynomial.
```

```python
#27 Can polynomial regression be applied to multiple variables?
#  Yes, polynomial regression can absolutely be applied to multiple variables. This is often referred to as multivariate polynomial regression.
```

```python
#28 What are the limitations of polynomial regression?
# While polynomial regression is a powerful tool for modeling non-linear relationships, it has several limitations that are important to consider:

# Overfitting: This is the most significant limitation. High-degree polynomials can create very complex curves that fit the training data almost perfectly,
# including the noise. This leads to poor generalization to new, unseen data. The model essentially memorizes the training data instead of learning the
# underlying relationship.

# Sensitivity to Outliers: Polynomial regression can be highly sensitive to outliers. A few extreme data points can significantly distort the shape of the
# fitted curve.

# Interpretability: As the degree of the polynomial increases, the coefficients become harder to interpret. It becomes difficult to explain the meaning of
# each term in the equation in a practical context.

# Extrapolation Issues: Extrapolating (making predictions outside the range of the observed data) can be extremely unreliable with polynomial regression.
# The curve can behave unpredictably outside the data range, leading to wildly inaccurate predictions.

# Computational Complexity: Fitting high-degree polynomials can be computationally expensive, especially with a large number of data points.

# Oscillations: High-degree polynomials can exhibit unwanted oscillations or "wiggles" between data points, even if there's no real reason for such oscillations.

# Choosing the Right Degree: Selecting the appropriate degree for the polynomial is crucial but can be challenging. Too low a degree will underfit the data
# (not capture the true relationship), while too high a degree will overfit.

# Multicollinearity: In multivariate polynomial regression (with multiple independent variables), creating polynomial terms and interaction terms can lead to
# high multicollinearity (high correlation between predictor variables). This can make it difficult to estimate the coefficients accurately.
```

```python
#29 What methods can be used to evaluate model fit when selecting the degree of a polynomial?
# The following methods can be used to evaluate model fit when selecting the degree of a polynomial:
# Statistical approaches are R squared(when r squared is close to one that means model fitted data well like 0.8 or 0.9), Adjusted R-Squared(It penalizes the
# the unnecessary addition of predictors.) and MSE or RMSE(It measures the error in model prediction and actual. Lower the value, better the model it is.)
# Visulization approach (Plotting the fitted curve: Plot the polynomial curve along with the actual data points. This gives a visual sense of how well the
# curve captures the underlying trend. Look for a curve that follows the data without excessive oscillations or fitting to noise.)
```

```python
# #30 Why is visualization important in polynomial regression?
# # Visualizations play a crucial role in polynomial regression for several reasons:

# 1. Understanding the Relationship:

# Non-linear patterns: Polynomial regression is used to model non-linear relationships between variables. Visualizations, such as scatter plots with the fitted polynomial curve, help to understand the nature of this non-linearity. You can see how the curve bends and fits the data points, revealing patterns that might not be apparent from numerical metrics alone.
# 2. Assessing Model Fit:

# Goodness of fit: By plotting the fitted curve against the actual data points, you can visually assess how well the model captures the underlying trend. A good fit will show the curve closely following the data points.
# Overfitting and underfitting: Visualizations can help identify overfitting (where the curve is too complex and fits noise in the data) and underfitting (where the curve is too simple and doesn't capture the trend). An overfit curve will often have excessive oscillations, while an underfit curve will not follow the data closely.
# 3. Choosing the Degree of the Polynomial:

# Optimal complexity: Visualizations can assist in selecting the appropriate degree for the polynomial. By plotting curves of different degrees, you can visually compare their fit and choose the one that balances capturing the trend without overfitting.
# 4. Identifying Outliers and Influential Points:

# Data anomalies: Visualizations can help identify outliers (data points far from the general trend) and influential points (data points that have a strong effect on the fitted curve). These points can affect the model and might require further investigation.
# 5. Communicating Results:

# Clear presentation: Visualizations provide a clear and intuitive way to communicate the results of polynomial regression to others, even those without a strong statistical background.
```

```python
#31 How is polynomial regression implemented in Python?
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score

#prepare the data
np.random.seed(0)  # for reproducibility
x = np.linspace(-3, 3, 100)
y = 2 * x**3 + 3 * x**2 - 5 * x + 1 + np.random.normal(0, 10, 100) # Add some noise
x = x.reshape(-1, 1) # Reshape x to be a 2D array, required by scikit-learn

degrees = [1, 2, 3, 4] # Test different degrees
plt.figure(figsize=(10, 8))
for i, degree in enumerate(degrees):
    plt.subplot(2, 2, i + 1)
    poly = PolynomialFeatures(degree=degree)
    x_poly = poly.fit_transform(x)
    model = LinearRegression()
    model.fit(x_poly, y)
    y_pred = model.predict(x_poly)
    mse = mean_squared_error(y, y_pred)
    r2 = r2_score(y, y_pred)
    plt.scatter(x, y, label="Actual data")
    plt.plot(x, y_pred, color='red', label=f"Polynomial Regression (degree={degree})\nMSE={mse:.2f}, R^2={r2:.2f}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(f"Degree {degree}")
    plt.legend()
plt.tight_layout()
plt.show()
```
